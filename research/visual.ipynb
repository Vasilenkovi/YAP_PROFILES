```python
import numpy as np
import pandas as pd
import umap.umap_ as umap
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
import plotly.express as px
from sklearn.preprocessing import StandardScaler

# Загрузка эмбеддингов
embeddings = np.load("embeddings_smart.npy")

# Стандартизация
scaler = StandardScaler()
embeddings_scaled = scaler.fit_transform(embeddings)

# Загрузка данных с кластерами
df_dbscan = pd.read_csv("articles_without_orcid_dbscan.csv")
df_kmeans = pd.read_csv("articles_without_orcid_kmeans.csv")

# Визуализация DBSCAN
print("\nВизуализация DBSCAN...")
reducer = umap.UMAP(random_state=42)
embeddings_2d = reducer.fit_transform(embeddings_scaled)

plt.figure(figsize=(12, 8))
cmap = ListedColormap(plt.cm.tab20.colors)
scatter = plt.scatter(
    embeddings_2d[:, 0],
    embeddings_2d[:, 1],
    c=df_dbscan["dbscan_cluster"].astype(int),
    cmap=cmap,
    s=10,
    alpha=0.6
)
plt.title("DBSCAN кластеризация (UMAP)")
plt.xlabel("UMAP 1")
plt.ylabel("UMAP 2")
plt.colorbar(scatter, label="Кластер")
plt.savefig("dbscan_clusters.png", dpi=200, bbox_inches='tight')
plt.close()

# Визуализация KMeans
print("\nВизуализация KMeans...")
plt.figure(figsize=(12, 8))
scatter = plt.scatter(
    embeddings_2d[:, 0],
    embeddings_2d[:, 1],
    c=df_kmeans["kmeans_cluster"].astype(int),
    cmap=cmap,
    s=10,
    alpha=0.6
)
plt.title(f"KMeans кластеризация (k={df_kmeans['kmeans_cluster'].nunique()})")
plt.xlabel("UMAP 1")
plt.ylabel("UMAP 2")
plt.colorbar(scatter, label="Кластер")
plt.savefig("kmeans_clusters.png", dpi=200, bbox_inches='tight')
plt.close()

# Сохранение интерактивных графиков в HTML
print("\nСохранение интерактивных графиков в HTML...")
fig_dbscan = px.scatter(
    df_dbscan,
    x=embeddings_2d[:, 0],
    y=embeddings_2d[:, 1],
    color=df_dbscan["dbscan_cluster"].astype(str),
    hover_name="title",
    title="DBSCAN кластеризация",
    hover_data=["dbscan_cluster"]
)
fig_dbscan.write_html("dbscan_clusters.html")

fig_kmeans = px.scatter(
    df_kmeans,
    x=embeddings_2d[:, 0],
    y=embeddings_2d[:, 1],
    color=df_kmeans["kmeans_cluster"].astype(str),
    hover_name="title",
    title=f"KMeans кластеризация (k={df_kmeans['kmeans_cluster'].nunique()})",
    hover_data=["kmeans_cluster"]
)
fig_kmeans.write_html("kmeans_clusters.html")

print("\nВсе результаты сохранены:")
print("- dbscan_clusters.png")
print("- kmeans_clusters.png")
print("- dbscan_clusters.html")
print("- kmeans_clusters.html")
```
